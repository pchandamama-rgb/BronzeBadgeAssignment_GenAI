```python
import matplotlib.pyplot as plt
```

```python
# Input Layer
inputs = [
    "TransactionAmount",
    "TransactionTime",
    "MerchantCategory",
    "CustomerAge",
    "AccountBalance",
    "TransactionsToday"
]
```

```python
fig, ax = plt.subplots(figsize=(10,6))
for i, label in enumerate(inputs):
    ax.scatter(1, i, s=800, color='skyblue')
    ax.text(-1, i, label, fontsize=10, fontweight='bold')

# Hidden Layer 1
for i in range(8):
    ax.scatter(3, i-1, s=600, color='orange')

# Hidden Layer 2
for i in range(4):
    ax.scatter(5, i+1, s=600, color='green')

# Output Layer
ax.scatter(7,2,s=900,color='red')
ax.text(7.3,2,"Fraud\n(0/1)",fontsize=10, fontweight='bold')

# Connections
for i in range(6):
    for j in range(8):
        ax.plot([1,3],[i,j-1],'gray',linewidth=0.5)

for i in range(8):
    for j in range(4):
        ax.plot([3,5],[i-1,j+1],'gray',linewidth=0.5)

for i in range(4):
    ax.plot([5,7],[i+1,2],'gray',linewidth=0.8)

ax.set_xlim(0,8)
ax.set_ylim(-2,7)
ax.axis('off')

plt.title("Neural Network Architecture for Credit Card Fraud Detection")
plt.show()
```
