PDF read

```python
import fitz

filename = list(uploaded.keys())[0]

doc = fitz.open(filename)

pages = []

for i, page in enumerate(doc):
    text = page.get_text()

    pages.append({
        "page": i + 1,
        "text": text
    })

print("Total Pages :", len(pages))
```

PDF first page Text

```python
print(pages[0]["text"][:1000])
```

Create LangChain Documents

```python
from langchain_core.documents import Document

documents = []

for p in pages:

    documents.append(
        Document(
            page_content=p["text"],
            metadata={
                "page": p["page"]
            }
        )
    )

print(len(documents))
```

Chunking

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = splitter.split_documents(documents)

print("Chunks :", len(chunks))
```

```python
print(chunks[0].page_content)
```

```python
print(chunks[0].metadata)
```

Embedding Model

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

Create FAISS Database

```python
from langchain_community.vectorstores import FAISS

db = FAISS.from_documents(
    chunks,
    embedding
)

print("Vector Database Created Successfully")
```

Test Similarity Search

```python
question = "What is waiting period?"

docs = db.similarity_search(
    question,
    k=1
)

for d in docs:
    print("Page:", d.metadata["page"])
    print(d.page_content[:500])
    print("=" * 80)
```

Ask Question Function

```python
def ask_policy(question):

    docs = db.similarity_search(question, k=4)

    context = ""

    pages = []

    for d in docs:

        context += d.page_content + "\n\n"

        pages.append(str(d.metadata["page"]))

    prompt = f"""
You are an Insurance Policy Expert.

Rules:

1. Answer ONLY from the policy context.

2. If answer is not available,
say

"I couldn't find this information in the uploaded policy."

3. Keep answer short.

4. Mention important limits, waiting periods and conditions.

Policy Context:

{context}

Question:

{question}

Answer:
"""
    response = llm.invoke(prompt)

    if isinstance(response.content, list):
        answer = ""

        for item in response.content:
            if isinstance(item, dict):
                answer += item.get("text", "")
            else:
                answer += str(item)

    else:
        answer = response.content

    return answer, sorted(set(pages))
```

```python
while True:
    q = input("Ask Question : ")

    if q.lower() == "exit":

        break

    ans, pages = ask_policy(q)

    print("\nAnswer\n")

    print(ans)

    print("\nSource Pages :", pages)

    print("="*80)
```
